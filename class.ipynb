```python
wt = float(input("enter total weight kg"))
ht = float(input("enter total height meter"))

bmi = wt / (ht ** 2)
print("BMI:", round(bmi, 2))

if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Normal weight")
elif bmi < 30:
    print("Overweight")
else:
    print("Obese")
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("add:", a + b)
print("sub:", a - b)
print("product:", a * b)
print("division:", a / b)
print("floor:", a // b)
print("modulus:", a % b)
print("power:", a ** b)

if a > b:
    print("greatest:", a)
else:
    print("greatest:", b)

if a < b:
    print("smallest:", a)
else:
    print("smallest:", b)

print("equal:", a == b)


print("and:", a & b)
print("or:", a | b)
print("xor:", a ^ b)
print("left:", a << b)
print("right:", a >> b)
```
