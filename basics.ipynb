```python
class BankAccount:
    def __init__(self,name,bal=1000):
        self.name = name
        self.bal = bal
        print("Curr_balance =",self.bal)
    def deposit(self):
        amnt = int(input("Enter deposit amount: "))
        self.bal += amnt
        print("deposit sucess !")
        print("REmainin bal = ",self.bal)
    def withdraw(self):
        amnt = int(input("Enter withdrawal amount: "))
        if(self.bal >= amnt):
            self.bal -= amnt
            print("withdraw sucess !")
            print("remaining balance =",self.bal)
        else:
            print("Not enough Balance !")

obj = BankAccount("Amruth")
obj.withdraw()
obj.deposit()
```

```python
class College:
    def __init__(self,name,mark):
        self.name = name
        self.mark = mark
        print("constroctor",self.name,self.mark)

    def display(self):
        print("Parent Class")

class Student(College):
    def __init__(self,place,name,mark):
        self.place = place
        College.__init__(self,name,mark)  # Super().__init__(name,mark)
        print("child const",self.place)

    def child_method(self):
        print("child class")

obj = Student("calicut","abc",85)
obj.display()
obj.child_method()
```

```python
#Polymorphism

class parott:
    def fly():
        print("can fly !")
    def swim():
        print("cant swim !")

class penquin:
    def fly():
        print("cant fly!")
    def swim():
        print("can swim !")

def test(bird):
    bird.fly()
    bird.swim()


obj = parott
test(obj)

obj1 = penquin
test(obj1)
```

```python
#overriding

class college:

    def add(self):
        print("Parent")

class student(college):

    def add(self):
        super().add()
        print("child")

obj = student()
obj.add()
```
