```python
#2. Print the first and last character of a string.

s = "stringso"
print(s[0],s[-1])
```

```python
# 3. Convert a string to uppercase and lowercase.

s = "alPHa"

u = s.upper()
l = s.lower()

print(u,l)
```

```python
# 4. Check whether a string contains the word "python"

s = "nopython"
if "python" in s:
    print("found")
else:
    print("not")
```

```python
#  5. Count how many times the letter "a" appears in a string.

s = "akhila"
print(s.count("a"))
```

```python
# 6. Count vowels in a given string
s = "icecream"
vow = "aeiou"
count = 0

for x in s:
    if x in vow:
        count +=1
print(count)
```

```python
#  7. Check whether a string is a palindrome.

s = "brurb"
c = s[::-1]

if s == c:
    print("true")
else:
    print("false")
```

```python
# 8. Replace all spaces with -.

s = "motor cycle"
f = s.replace(" ","-")

print(f)
```

```python
# 9. Print each character of a string on a new line using a loop.

s = "new string"
for x in s:
    print(x)
```

```python
# 10. If a string length is more than 5, print "Long string", else "Short string".
def long():
    print("long")

def short():
    print("short")

def length():
    s = input("String")
    if len(s) > 5:
        long()
    else:
        short()

length()
```
