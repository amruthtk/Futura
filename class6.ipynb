```python
class BankAccount:
    def __init__(self, name, bal=0):
        self.name = name
        self.bal = bal
        print("Current balance =", self.bal)

    def deposit(self):
        amnt = int(input("Enter deposit amount: "))
        self.bal += amnt
        print("Deposit success!")
        print("Remaining balance =", self.bal)

    def withdraw(self):
        amnt = int(input("Enter withdrawal amount: "))
        if self.bal >= amnt:
            self.bal -= amnt
            print("Withdraw success!")
            print("Remaining balance =", self.bal)
        else:
            print("Not enough Balance!")


obj = BankAccount("Amruth", 1000)
obj.withdraw()
obj.deposit()
```

```python
try:
    balance = float(input("Enter your balance: "))

    try:
        withdrawal = float(input("Enter withdrawal amount: "))

        if withdrawal > balance:
            raise Exception("Withdrawal amount exceeds balance.")

        balance = balance - withdrawal

        print("Withdrawal successful!")
        print("Remaining balance:", balance)

    except ValueError:
        print("Invalid withdrawal amount. Please enter a number.")

    except Exception as e:
        print("Error:", e)

except ValueError:
    print("Invalid balance. Please enter a number.")
```

```python
class Student:
    school_name = "ABC School"

    def __init__(self, name, mark1, mark2, mark3):
        self.name = name
        self.mark1 = mark1
        self.mark2 = mark2
        self.mark3 = mark3


    def calculate_total(self):
        return self.mark1 + self.mark2 + self.mark3

    @classmethod
    def change_school_name(cls, new_name):
        cls.school_name = new_name

    @staticmethod
    def check_pass(mark):
        return mark >= 40

student1 = Student("Anu", 50, 60, 70)


print("Student Name:", student1.name)
print("Total Marks:", student1.calculate_total())

Student.change_school_name("XYZ School")
print("School Name:", Student.school_name)

print("Pass:", Student.check_pass(50))
```

```python
class Circle:
    pi = 3.14

    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return Circle.pi * self.radius * self.radius

    @classmethod
    def change_pi(cls, new_pi):
        cls.pi = new_pi

    @staticmethod
    def check_radius(radius):
        return radius > 0

c1 = Circle(5)

print("Radius is positive:", Circle.check_radius(c1.radius))

print("Area:", c1.calculate_area())

Circle.change_pi(3.14159)

print("New Area:", c1.calculate_area())
```

```python
class BankAccount:
    interest_rate = 5.0

    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
        print("Balance:", self.balance)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
            print("Balance:", self.balance)
        else:
            print("Insufficient balance")

    @classmethod
    def change_interest_rate(cls, new_rate):
        cls.interest_rate = new_rate
        print("New interest rate:", cls.interest_rate, "%")


    @staticmethod
    def validate_account_number(account_number):
        return len(str(account_number)) == 6 and str(account_number).isdigit()
account = BankAccount(123456, 10000)

account.deposit(2000)
account.withdraw(3000)

BankAccount.change_interest_rate(6.5)

print(BankAccount.validate_account_number(123456))
print(BankAccount.validate_account_number(12345))
```

```python
class Student:
    school_name = "ABC School"

    def __init__(self, id, name, age, course, marks):
        self.id = id
        self.name = name
        self.age = age
        self.course = course
        self.marks = marks

    def display(self):
        print(self.id, self.name, self.age, self.course, self.marks, Student.school_name)

    def grade(self):
        if self.marks >= 90:
            print("A")
        elif self.marks >= 75:
            print("B")
        elif self.marks >= 50:
            print("C")
        else:
            print("Fail")


    def update_marks(self, marks):
        self.marks = marks

    @classmethod
    def change_school(cls, name):
        cls.school_name = name

    @staticmethod
    def pass_fail(marks):
        return "Pass" if marks >= 50 else "Fail"


s1 = Student(101, "Anu", 18, "CS", 92)
s2 = Student(102, "Rahul", 19, "AI", 78)
s3 = Student(103, "Meera", 18, "CE", 45)

s1.display()
s2.display()
s3.display()

print(s1.name, s1.grade())
print(s2.name, s2.grade())
print(s3.name, s3.grade())


s3.update_marks(65)
print("Updated:", s3.name, s3.marks, s3.grade())


Student.change_school("XYZ College")
s1.display()

print(Student.pass_fail(s1.marks))
print(Student.pass_fail(s2.marks))
print(Student.pass_fail(s3.marks))
```
